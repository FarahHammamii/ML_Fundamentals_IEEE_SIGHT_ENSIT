**Q4. A value of 45 study hours/week is unusual. What should you do first?**

<details>
<summary><b>Reveal 2 </b></summary>

**Q3: C. Q4: C.**

EDA is not just `head()`, `describe()`, or a histogram. Those are tools used to ask:

- What do the variables actually mean?
- What is typical or unusual?
- What is missing?
- Are duplicates real or accidental?
- Could the data itself mislead us?

Additional information arrives: one source recorded **study hours per day**, while the expected unit was **hours per week**.

The outlier was therefore a clue about the data-generating process.

**Important:** an outlier is not automatically “bad data.” It can be a typo, a unit mismatch, or an important observation.
</details>

<details>
<summary><b>Reveal 3B — leakage</b></summary>

**B — Do not use it.**

The feature is highly predictive but invalid for this use case because it comes from the future relative to the prediction moment.

This is **data leakage**.

A powerful feature can make the experiment look excellent while making the deployed system impossible or misleading.

This is why understanding the real-world process comes before optimizing a model.
</details>

# Checkpoint 4 — A higher score can be a worse experiment

Two analysts report results on unseen test students.

| Experiment | Features available to model | Test accuracy |
|---|---|---:|
| Model A | Study + attendance + assignments + **final exam score** | **97%** |
| Model B | Study + attendance + assignments | **82%** |

### Vote

**Q6. Which experiment is valid for the stated deployment scenario?**

A. Model A because 97% > 82%  
B. Model B because its inputs exist at prediction time  
C. Both are equally valid  
D. Neither, because accuracy can never be used

Then answer in one sentence:

> Why can a lower score represent a better ML experiment?

<details>
<summary><b>Reveal 4</b></summary>

**B — Model B is the valid experiment for this scenario.**

The 82% result is less impressive numerically, but it estimates performance under conditions closer to actual deployment.

This is the difference between **optimizing an observed number** and asking whether the learned relationship can **generalize** to the real situation.

A trustworthy ML experiment respects how the model will actually receive information.
</details>

# Checkpoint 5 — Train ≠ test

Three models are evaluated.

| Model | Training accuracy | Test accuracy |
|---|---:|---:|
| A | 70% | 69% |
| B | 86% | **84%** |
| C | **100%** | 68% |

### Interpret, don't calculate

**Q7. Match each model to the most likely situation:**

- Underfitting
- Better fit / better generalization
- Overfitting

### Q8.
Model C has perfect training accuracy. Did it therefore learn best?

A. Yes  
B. No

Explain your answer before revealing.

<details>
<summary><b>Reveal 5</b></summary>

A → **Underfitting**  
B → **Better fit / better generalization among these three examples**  
C → **Overfitting**

Model C demonstrates why training performance is not the final objective.

A model can memorize the sample it saw and still perform poorly on new observations.

Think of the course analogy:

- **Train set** = solved exercises used for learning
- **Test set** = a new exam question that remained untouched

The real goal is performance on unseen data.
</details>
